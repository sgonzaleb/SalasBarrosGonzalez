```python
# ==========================================
# ENTREGA 03 - VISUALIZACIÓN ATÓMICA CON ALTAIR
# Alumno: Santiago Barros
# Tema: El doble rasero hacia las promesas juveniles vs delanteros consagrados
# ==========================================

import pandas as pd
import altair as alt

# 1. Cargar o estructurar los datos del análisis
data = {
    'Jugador': ['Floyd Samba (17 años)', 'Juveniles Sub-21', 'Cody Gakpo', 'Joshua Zirkzee'],
    'Tipo': ['Promesa / Juvenil', 'Promesa / Juvenil', 'Consagrado / Primer Equipo', 'Consagrado / Primer Equipo'],
    'Equipo': ['Manchester City', 'Manchester United', 'Liverpool', 'Manchester United'],
    'Total_Menciones': [11, 14, 10, 8],
    'Tono_Predominante': ['Respaldo e Ilusión (100%)', 'Respaldo e Ilusión (93%)', 'Resistencia / Disculpas tardías (70%)', 'Crítica e Impaciencia (88%)'],
    'Porcentaje_Aprobacion': [100, 93, 30, 12]
}

df_vis = pd.DataFrame(data)
df_vis.to_csv('base_delanteros_vis.csv', index=False)
print("Base CSV exportada exitosamente.")

# 2. Generación de la visualización interactiva con Altair
bars = alt.Chart(df_vis).mark_bar(cornerRadiusTopRight=5, cornerRadiusBottomRight=5).encode(
    x=alt.X('Porcentaje_Aprobacion:Q',
            title='Tolerancia y Aprobación del Hincha (%)',
            scale=alt.Scale(domain=[0, 115])),
    y=alt.Y('Jugador:N',
            sort='-x',
            title='Atacante / Referente evaluado'),
    color=alt.Color('Tipo:N',
                    scale=alt.Scale(domain=['Promesa / Juvenil', 'Consagrado / Primer Equipo'],
                                    range=['#27ae60', '#c0392b']),
                    title='Categoría del atacante'),
    tooltip=[
        alt.Tooltip('Jugador:N', title='Jugador'),
        alt.Tooltip('Equipo:N', title='Club'),
        alt.Tooltip('Tipo:N', title='Categoría'),
        alt.Tooltip('Porcentaje_Aprobacion:Q', title='Aprobación (%)'),
        alt.Tooltip('Tono_Predominante:N', title='Tono en comentarios'),
        alt.Tooltip('Total_Menciones:Q', title='Menciones analizadas')
    ]
).properties(
    title={
        "text": ["El doble rasero en ataque: La ceguera hacia las promesas vs el juicio al consagrado"],
        "subtitle": [
            "Porcentaje de respaldo y tolerancia en comentarios de Instagram en partidos del Big Six",
            "Muestra: Comentarios de Liverpool, Manchester City y Manchester United"
        ],
        "fontSize": 14,
        "subtitleFontSize": 11
    },
    width=600,
    height=260
)

# Etiquetas de texto sobre las barras
text = bars.mark_text(
    align='left',
    baseline='middle',
    dx=6,
    fontSize=11,
    fontWeight='bold'
).encode(
    text=alt.Text('Porcentaje_Aprobacion:Q', format='.0f')
)

chart_final = bars + text

# 3. Exportar a formato HTML interactivo para webstory
chart_final.save('vis_01.html')
print("Archivo vis_01.html generado.")
chart_final
```
